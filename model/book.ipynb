```python

import pandas as pd
import requests
from io import StringIO

CHANNEL_ID = 3171928

url = (
    f"https://api.thingspeak.com/channels/{3171928}/feeds.csv"
    "?results=8000"
)

response = requests.get(url, timeout=30)
response.raise_for_status()

df = pd.read_csv(StringIO(response.text))

# Rename the confirmed sensor fields
rename_map = {
    "field1": "flame",
    "field2": "mq2_smoke",
    "field3": "mq7_co",
    "field4": "temperature",
    "field5": "humidity"
}

df = df.rename(columns=rename_map)

# Convert sensor columns to numeric values
sensor_cols = [
    "flame", "mq2_smoke", "mq7_co",
    "temperature", "humidity"
]

for col in sensor_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print("Dataset shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nMissing values:")
print(df[sensor_cols].isnull().sum())

print("\nSensor summary:")
display(df[sensor_cols].describe())

print("\nFirst 10 rows:")
display(df.head(10))

df.to_csv("efireguard_sensor_data.csv", index=False)
print("\nSaved to efireguard_sensor_data.csv")
```

```python

import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix,
    classification_report
)

# 1. Load the dataset without MQ-7
file_path = r"F:\ML\data\efireguard_3sensor_dataset.csv"
df = pd.read_csv(file_path)

# 2. Keep only the required five columns
required_columns = [
    "flame",
    "mq2_smoke",
    "temperature",
    "humidity",
    "target"
]

df = df[required_columns].copy()

# 3. Keep valid binary target labels
df = df[df["target"].isin([0, 1])].copy()

# 4. Use sensor readings as inputs
# Do NOT include flame because target is derived from flame
features = ["mq2_smoke", "temperature", "humidity"]

df = df.dropna(subset=features + ["target"])

X = df[features]
y = df["target"].astype(int)

print("Dataset shape:", df.shape)
print("\nTarget distribution:")
print(y.value_counts())

# 5. Split the data, preserving class proportions
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 6. Create and train Random Forest
model = make_pipeline(
    SimpleImputer(strategy="median"),
    RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
)

model.fit(X_train, y_train)

# 7. Evaluate the trained model
pred = model.predict(X_test)

print("\nAccuracy:",
      round(accuracy_score(y_test, pred), 4))

print("Balanced accuracy:",
      round(balanced_accuracy_score(y_test, pred), 4))

print("\nConfusion matrix:")
print(confusion_matrix(y_test, pred, labels=[0, 1]))

print("\nClassification report:")
print(classification_report(
    y_test,
    pred,
    labels=[0, 1],
    target_names=["Flame state 0", "Flame state 1"],
    zero_division=0
))

# 8. Save the model
artifact = {
    "model": model,
    "features": features,
    "model_type": "RandomForestClassifier",
    "target_description": "Recorded flame sensor state"
}

output_path = r"F:\ML\models\efireguard_rf_3sensor.joblib"
joblib.dump(artifact, output_path)

print("\nModel saved at:", output_path)
print("Input features:", features)
```

```python

import pandas as pd

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    GridSearchCV
)
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report
)

# Load dataset
df = pd.read_csv(
    r"F:\ML\data\efireguard_3sensor_dataset.csv"
)

features = ["mq2_smoke", "temperature", "humidity"]
df = df.dropna(subset=features + ["target"])
df = df[df["target"].isin([0, 1])]

X = df[features]
y = df["target"].astype(int)

# Keep a test set separate from model tuning
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Random Forest with tunable complexity
pipeline = make_pipeline(
    SimpleImputer(strategy="median"),
    RandomForestClassifier(
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )
)

# Regularization search
param_grid = {
    "randomforestclassifier__max_depth": [5, 10, 15],
    "randomforestclassifier__min_samples_split": [5, 10, 20],
    "randomforestclassifier__min_samples_leaf": [2, 5, 10],
    "randomforestclassifier__max_features": ["sqrt", 0.8]
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

search = GridSearchCV(
    pipeline,
    param_grid,
    cv=cv,
    scoring="balanced_accuracy",
    return_train_score=True,
    n_jobs=-1,
    refit=True
)

search.fit(X_train, y_train)

best_model = search.best_estimator_
best_index = search.best_index_

print("Best parameters:")
print(search.best_params_)

print("\nMean CV training score:",
      round(search.cv_results_["mean_train_score"][best_index], 4))

print("Mean CV validation score:",
      round(search.best_score_, 4))

# Evaluate on held-out test data
pred = best_model.predict(X_test)

print("\nTest accuracy:",
      round(accuracy_score(y_test, pred), 4))

print("Test balanced accuracy:",
      round(balanced_accuracy_score(y_test, pred), 4))

print("\nClassification report:")
print(classification_report(y_test, pred, zero_division=0))
```

```python
import pandas as pd
import numpy as np
import joblib
import json
from pathlib import Path

from sklearn.model_selection import (
    train_test_split, StratifiedKFold, GridSearchCV
)
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score,
    precision_score, recall_score, f1_score,
    confusion_matrix, classification_report,
    roc_auc_score, average_precision_score
)

# =====================================================
# 1. LOAD DATA
# =====================================================

DATA_PATH = r"F:\ML\data\efireguard_binary_provisional.csv"

df = pd.read_csv(DATA_PATH)

features = ["mq2_smoke", "temperature", "humidity"]
target = "target"

X = df[features].apply(pd.to_numeric, errors="coerce")
y = pd.to_numeric(df[target], errors="coerce")

data = X.copy()
data[target] = y
data = data.dropna()

X = data[features]
y = data[target].astype(int)

if set(y.unique()) != {0, 1}:
    raise ValueError("Target must contain classes 0 and 1.")

print("Dataset size:", len(data))
print("Class distribution:\n", y.value_counts())

# =====================================================
# 2. SPLIT DATA
# =====================================================

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    stratify=y,
    random_state=42
)

# =====================================================
# 3. STRONGER REGULARIZATION SEARCH
# =====================================================

rf = RandomForestClassifier(
    random_state=42,
    n_jobs=-1
)

param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [1, 2, 3, 4],
    "min_samples_split": [30, 50, 80, 120],
    "min_samples_leaf": [20, 30, 50, 80],
    "max_features": ["sqrt", 0.5, 1.0],
    "class_weight": [None, "balanced"]
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    scoring={
        "accuracy": "accuracy",
        "balanced_accuracy": "balanced_accuracy",
        "precision": "precision",
        "recall": "recall",
        "f1": "f1",
        "roc_auc": "roc_auc"
    },
    # First fit the candidates; select below from CV results.
    refit=False,
    cv=cv,
    n_jobs=-1,
    return_train_score=True,
    verbose=1
)

search.fit(X_train, y_train)

# =====================================================
# 4. CHOOSE USING TRAINING-SPLIT CV ONLY
# =====================================================

results = pd.DataFrame(search.cv_results_)

eligible = results[
    results["mean_test_accuracy"] < 0.90
].copy()

if eligible.empty:
    # No candidate reached the requested CV target.
    # Choose the lowest-CV-accuracy candidate and disclose that fact.
    selected = results.loc[
        results["mean_test_accuracy"].idxmin()
    ]
    print("\nNo candidate achieved CV accuracy below 90%.")
else:
    # Closest to 89%, with higher accuracy used as a tie-breaker.
    eligible["distance_from_89"] = (
        eligible["mean_test_accuracy"] - 0.89
    ).abs()

    selected = eligible.sort_values(
        ["distance_from_89", "mean_test_accuracy"],
        ascending=[True, False]
    ).iloc[0]

best_params = selected["params"]

print("\nSelected parameters:")
print(best_params)

print(
    "Training CV accuracy:",
    f"{selected['mean_test_accuracy'] * 100:.2f}%"
)
print(
    "Training CV accuracy standard deviation:",
    f"{selected['std_test_accuracy'] * 100:.2f}%"
)
print(
    "Training CV balanced accuracy:",
    f"{selected['mean_test_balanced_accuracy'] * 100:.2f}%"
)

# =====================================================
# 5. FIT SELECTED MODEL
# =====================================================

model = RandomForestClassifier(
    random_state=42,
    n_jobs=-1,
    **best_params
)

model.fit(X_train, y_train)

# =====================================================
# 6. FINAL HELD-OUT TEST EVALUATION
# =====================================================

y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

tn, fp, fn, tp = confusion_matrix(
    y_test, y_pred, labels=[0, 1]
).ravel()

accuracy = accuracy_score(y_test, y_pred)
balanced_acc = balanced_accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

specificity = tn / (tn + fp) if tn + fp else 0
fpr = fp / (fp + tn) if fp + tn else 0
fnr = fn / (fn + tp) if fn + tp else 0

roc_auc = roc_auc_score(y_test, y_prob)
avg_precision = average_precision_score(y_test, y_prob)

print("\n========== ACTUAL TEST METRICS ==========")
print(f"Accuracy:           {accuracy * 100:.2f}%")
print(f"Balanced accuracy:  {balanced_acc:.4f}")
print(f"Precision:          {precision:.4f}")
print(f"Recall:             {recall:.4f}")
print(f"Specificity:        {specificity:.4f}")
print(f"F1-score:           {f1:.4f}")
print(f"False positive rate:{fpr:.4f}")
print(f"False negative rate:{fnr:.4f}")
print(f"ROC-AUC:            {roc_auc:.4f}")
print(f"Average precision:  {avg_precision:.4f}")

print("\nConfusion matrix [[TN, FP], [FN, TP]]:")
print([[tn, fp], [fn, tp]])

print("\nClassification report:")
print(classification_report(
    y_test, y_pred,
    labels=[0, 1],
    zero_division=0
))

# =====================================================
# 7. SAVE RESULTS WITHOUT OVERWRITING OLD MODEL
# =====================================================

output_dir = Path(r"F:\ML\evaluation_below_90_v2")
output_dir.mkdir(parents=True, exist_ok=True)

joblib.dump(
    {
        "model": model,
        "features": features,
        "target_description": "Recorded flame sensor state"
    },
    output_dir / "random_forest_candidate_v2.joblib"
)

metrics = {
    "selected_parameters": best_params,
    "cv_accuracy": float(selected["mean_test_accuracy"]),
    "test_accuracy": float(accuracy),
    "balanced_accuracy": float(balanced_acc),
    "precision": float(precision),
    "recall": float(recall),
    "specificity": float(specificity),
    "f1_score": float(f1),
    "false_positive_rate": float(fpr),
    "false_negative_rate": float(fnr),
    "roc_auc": float(roc_auc),
    "average_precision": float(avg_precision),
    "tn": int(tn),
    "fp": int(fp),
    "fn": int(fn),
    "tp": int(tp)
}

with open(output_dir / "metrics.json", "w") as file:
    json.dump(metrics, file, indent=4)

results.to_csv(
    output_dir / "all_cv_results.csv",
    index=False
)

print("\nSaved new candidate and results to:", output_dir)

if accuracy < 0.90:
    print("Measured test accuracy is below 90% on this split.")
else:
    print("Measured test accuracy is still 90% or higher.")
    print("Do not tune repeatedly against this test set.")
```

```python
"""
Fire/Smoke Detection - Evaluation Report Generator

Creates:
- Confusion matrix
- ROC curve with ROC-AUC
- Precision-Recall curve with Average Precision
- Metrics bar chart
- Class distribution plot
- Feature importance plot (when supported)
- CSV and TXT evaluation reports

IMPORTANT:
1. Set DATA_PATH to your original dataset CSV.
2. Set MODEL_PATH to the exact saved model file.
3. Set TARGET_COL to your target column.
4. If you already have a saved test set, use it instead of re-splitting
   the original dataset so the results match your previous evaluation.
"""

from pathlib import Path
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report, roc_curve, roc_auc_score,
    precision_recall_curve, average_precision_score, matthews_corrcoef,
    cohen_kappa_score, log_loss, brier_score_loss
)

# ----------------------- CONFIGURATION -----------------------
DATA_PATH = Path(r"F:\ML\data\efireguard_binary_provisional.csv")
EVAL_DIR = Path(r"F:\ML\evaluation_below_90_v2")

# IMPORTANT: Change this to the exact .joblib model filename in the folder.
# Example: Path(r"F:\ML\evaluation_below_90_v2\best_candidate.joblib")
MODEL_PATH = Path(r"F:\ML\evaluation_below_90_v2\best_candidate.joblib")

TARGET_COL = "target"
TEST_SIZE = 0.20
RANDOM_STATE = 42
POSITIVE_LABEL = 1
DECISION_THRESHOLD = 0.30  # Chosen from your validation threshold sweep
# ------------------------------------------------------------


def load_model(path):
    if not path.exists():
        candidates = sorted(EVAL_DIR.glob("*.joblib")) + sorted(EVAL_DIR.glob("*.pkl"))
        if len(candidates) == 1:
            path = candidates[0]
            print(f"Using model file: {path}")
        else:
            raise FileNotFoundError(
                f"Model file not found: {path}\n"
                "Set MODEL_PATH to the exact .joblib/.pkl filename in EVAL_DIR.\n"
                f"Found candidates: {[p.name for p in candidates]}"
            )

    artifact = joblib.load(path)
    if isinstance(artifact, dict):
        model = artifact.get("model")
        if model is None:
            model = artifact.get("estimator")
        if model is None:
            model = artifact.get("classifier")
        if model is None:
            raise ValueError(
                "The loaded dictionary does not contain 'model', 'estimator', "
                "or 'classifier'. Inspect the artifact keys and update load_model()."
            )
    else:
        model = artifact
    return model, path


def get_positive_probability(model, X, positive_label=1):
    if not hasattr(model, "predict_proba"):
        raise TypeError(
            "The saved model does not support predict_proba(), required for ROC/PR curves."
        )
    classes = list(model.classes_)
    if positive_label not in classes:
        raise ValueError(f"Positive label {positive_label} not found in model classes: {classes}")
    positive_index = classes.index(positive_label)
    return model.predict_proba(X)[:, positive_index]


def save_plot(filename):
    plt.tight_layout()
    plt.savefig(EVAL_DIR / filename, dpi=300, bbox_inches="tight")
    plt.close()
    print(f"Saved plot: {EVAL_DIR / filename}")


def main():
    EVAL_DIR.mkdir(parents=True, exist_ok=True)

    if not DATA_PATH.exists():
        raise FileNotFoundError(f"Dataset not found: {DATA_PATH}")

    df = pd.read_csv(DATA_PATH)
    if TARGET_COL not in df.columns:
        raise ValueError(f"Target column '{TARGET_COL}' not found. Columns: {list(df.columns)}")

    # Match the model's trained feature list where available.
    model, model_file = load_model(MODEL_PATH)
    if hasattr(model, "feature_names_in_"):
        feature_cols = list(model.feature_names_in_)
    else:
        feature_cols = [c for c in df.columns if c != TARGET_COL]

    missing = [c for c in feature_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Dataset is missing model feature columns: {missing}")

    X = df[feature_cols].copy()
    y = df[TARGET_COL].copy()

    # For numeric sensor columns, fill missing values with training-data medians.
    # If your model was trained with a Pipeline/imputer, that pipeline handles this itself.
    for col in X.columns:
        if X[col].isna().any():
            if pd.api.types.is_numeric_dtype(X[col]):
                X[col] = X[col].fillna(X[col].median())
            else:
                X[col] = X[col].fillna(X[col].mode().iloc[0])

    # Stratified split. Replace this block with your original saved test indices/data
    # if you need exact reproducibility with previous reported scores.
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=y
    )

    y_score = get_positive_probability(model, X_test, POSITIVE_LABEL)
    # Apply the tuned operating threshold consistently to the reported metrics.
    y_pred = (y_score >= DECISION_THRESHOLD).astype(int)

    labels = sorted(pd.unique(y))
    if POSITIVE_LABEL not in labels:
        raise ValueError(f"Positive label {POSITIVE_LABEL} is not in target values {labels}")

    cm = confusion_matrix(y_test, y_pred, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()
    specificity = tn / (tn + fp) if (tn + fp) else 0.0
    fpr_value = fp / (fp + tn) if (fp + tn) else 0.0
    fnr_value = fn / (fn + tp) if (fn + tp) else 0.0

    metrics = {
        "model_file": str(model_file),
        "dataset_file": str(DATA_PATH),
        "test_rows": len(y_test),
        "decision_threshold": DECISION_THRESHOLD,
        "accuracy": accuracy_score(y_test, y_pred),
        "balanced_accuracy": balanced_accuracy_score(y_test, y_pred),
        "precision": precision_score(y_test, y_pred, pos_label=POSITIVE_LABEL, zero_division=0),
        "recall_sensitivity": recall_score(y_test, y_pred, pos_label=POSITIVE_LABEL, zero_division=0),
        "specificity": specificity,
        "f1_score": f1_score(y_test, y_pred, pos_label=POSITIVE_LABEL, zero_division=0),
        "false_positive_rate": fpr_value,
        "false_negative_rate": fnr_value,
        "roc_auc": roc_auc_score(y_test, y_score),
        "average_precision": average_precision_score(y_test, y_score),
        "matthews_corrcoef": matthews_corrcoef(y_test, y_pred),
        "cohen_kappa": cohen_kappa_score(y_test, y_pred),
        "brier_score": brier_score_loss(y_test, y_score),
    }

    # Save numeric metrics and detailed class report.
    pd.DataFrame([metrics]).to_csv(EVAL_DIR / "evaluation_metrics.csv", index=False)
    with open(EVAL_DIR / "evaluation_report.txt", "w", encoding="utf-8") as f:
        f.write("FIRE AND SMOKE DETECTION - MODEL EVALUATION REPORT\n")
        f.write("=" * 58 + "\n\n")
        for key, value in metrics.items():
            if isinstance(value, (float, np.floating)):
                f.write(f"{key}: {value:.4f}\n")
            else:
                f.write(f"{key}: {value}\n")
        f.write("\nClassification report:\n")
        f.write(classification_report(y_test, y_pred, zero_division=0))
        f.write("\nConfusion matrix (labels [0, 1]):\n")
        f.write(np.array2string(cm))
    print("\nEvaluation metrics:")
    for key, value in metrics.items():
        print(f"{key}: {value:.4f}" if isinstance(value, (float, np.floating)) else f"{key}: {value}")
    print("\n", classification_report(y_test, y_pred, zero_division=0))

    # 1. Confusion matrix
    plt.figure(figsize=(6.5, 5.5))
    plt.imshow(cm, interpolation="nearest")
    plt.title("Confusion Matrix")
    plt.colorbar(label="Number of samples")
    plt.xticks([0, 1], ["Predicted 0", "Predicted 1"])
    plt.yticks([0, 1], ["Actual 0", "Actual 1"])
    for i in range(2):
        for j in range(2):
            plt.text(j, i, f"{cm[i, j]}", ha="center", va="center", fontsize=13)
    plt.xlabel("Predicted label")
    plt.ylabel("True label")
    save_plot("01_confusion_matrix.png")

    # 2. ROC curve
    fpr, tpr, _ = roc_curve(y_test, y_score, pos_label=POSITIVE_LABEL)
    auc_value = roc_auc_score(y_test, y_score)
    plt.figure(figsize=(7, 6))
    plt.plot(fpr, tpr, label=f"ROC curve (AUC = {auc_value:.3f})")
    plt.plot([0, 1], [0, 1], linestyle="--", label="Random classifier")
    plt.xlim(0, 1)
    plt.ylim(0, 1.02)
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate (Recall)")
    plt.title("Receiver Operating Characteristic (ROC) Curve")
    plt.legend(loc="lower right")
    plt.grid(True, alpha=0.3)
    save_plot("02_roc_curve.png")

    # 3. Precision-Recall curve
    precision_values, recall_values, _ = precision_recall_curve(
        y_test, y_score, pos_label=POSITIVE_LABEL
    )
    ap_value = average_precision_score(y_test, y_score)
    baseline = (y_test == POSITIVE_LABEL).mean()
    plt.figure(figsize=(7, 6))
    plt.plot(recall_values, precision_values, label=f"PR curve (AP = {ap_value:.3f})")
    plt.axhline(baseline, linestyle="--", label=f"Positive prevalence = {baseline:.3f}")
    plt.xlim(0, 1)
    plt.ylim(0, 1.02)
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("Precision–Recall Curve")
    plt.legend(loc="best")
    plt.grid(True, alpha=0.3)
    save_plot("03_precision_recall_curve.png")

    # 4. Metrics bar chart (unitless rates, all on 0-1 scale)
    rate_keys = [
        "accuracy", "balanced_accuracy", "precision", "recall_sensitivity",
        "specificity", "f1_score", "roc_auc", "average_precision"
    ]
    plt.figure(figsize=(10, 5.5))
    vals = [metrics[k] for k in rate_keys]
    bars = plt.bar([k.replace("_", "\n") for k in rate_keys], vals)
    plt.ylim(0, 1.05)
    plt.ylabel("Score")
    plt.title("Model Evaluation Metrics")
    plt.xticks(rotation=0, fontsize=8)
    plt.bar_label(bars, fmt="%.3f", padding=3)
    plt.grid(axis="y", alpha=0.3)
    save_plot("04_metrics_comparison.png")

    # 5. Class distribution
    counts = y.value_counts().sort_index()
    plt.figure(figsize=(6.5, 5))
    bars = plt.bar([str(i) for i in counts.index], counts.values)
    plt.xlabel("Target class")
    plt.ylabel("Number of rows")
    plt.title("Dataset Class Distribution")
    plt.bar_label(bars, padding=3)
    save_plot("05_class_distribution.png")

    # 6. Feature importance if the estimator exposes it.
    # For pipelines, inspect the final estimator.
    estimator = model
    if hasattr(model, "steps") and model.steps:
        estimator = model.steps[-1][1]

    if hasattr(estimator, "feature_importances_"):
        importances = np.asarray(estimator.feature_importances_)
        names = feature_cols
        if len(importances) == len(names):
            order = np.argsort(importances)
            plt.figure(figsize=(8, max(4, len(names) * 0.6)))
            bars = plt.barh(np.array(names)[order], importances[order])
            plt.xlabel("Feature importance")
            plt.title("Random Forest Feature Importance")
            plt.bar_label(bars, fmt="%.3f", padding=3)
            save_plot("06_feature_importance.png")
        else:
            print("Skipping feature importance: feature count mismatch.")
    else:
        print("Model does not expose feature_importances_; skipping this plot.")

    print(f"\nDone. Report files and plots saved in:\n{EVAL_DIR}")


if __name__ == "__main__":
    warnings.filterwarnings("default")
    main()
```
